Lets import data in our dataframe variable ,The most important part of the Pandas library is the DataFrame. A DataFrame holds the type of data you might think of as a table. This is similar to a sheet in Excel, or a table in a SQL database.



```python
import pandas as pd

# save filepath to variable for easier access
melbourne_data_path = '/Users/lawliet/Documents/college/LearningPython/melb_data.csv'

# read the data and store data in DataFrame titled melbourne_data
melbourne_data = pd.read_csv(melbourne_data_path)

#print a few top entries of data
print(f"We can display the whole data using describe.() function , But since it's too big we will use head function to get first few entries \n {melbourne_data.head()}")
```

## TO choose which variable we will be predicting we will need to nicely lay out all the columns , which we can easily do using the column function

```python
melbourne_data.columns
```

## Now that we have all of our columns we can select our prediction target , but before that we will take the easiest route to clean our data and get rid of all the data with no entries . Finally we can set out prediction value on y 

```python
#we will drop all the houses with missing dropna , na is 'not available'
melbourne_data = melbourne_data.dropna(axis=0)

#now we can set our prediction value in our y variable 
y = melbourne_data.Price
```

## Now that the prediction value have been selected , we will also select our features which we will use to train the model

```python
#first we create a list of features for easy access , but we can only use columns with int or float values NO STRINGS coz treee regressor can't do it 
features = ['Rooms', 'Distance', 'Postcode', 'Bedroom2', 'Bathroom', 'Car','Landsize', 'BuildingArea', 'YearBuilt', 'Lattitude','Longtitude', 'Propertycount']

#Then we create our X variable with all of these columns from our dataframe
X = melbourne_data[features]

#We can check if our features data is correct by using head 
print(f"muhehehe it works look at my fucking features data \n {X.head()}")
```

## Now that we have both our variables ready we will need to create a model , for this we will use a basic descision tree regressor model from ski learn 

```python
#importing the DecisionTreeRegressor from sklearn's tree folder that why we add .tree at the end 
from sklearn.tree import DecisionTreeRegressor

# Define model. Specify a number for random_state to ensure same results each run
melbourne_model = DecisionTreeRegressor(random_state=1)

#now we fit the model with our features dataframe (X) and predictions data frame (y)
melbourne_model.fit(X,y)
```

## Now that we have our model we can predict shi , we will use the head function to predict price(y) for the first few entries of X

```python
print(f'Predict price for these entries: \n{X.head()}')
print(f'\n Predictions are : {melbourne_model.predict(X.head())} \nWHEHEHEHHEHE LESSS GOOO I DID IT BITCH')
```

## This was a good test run but we literally tested our model on the data it was trained on so ofc it'll be almost correct.
 
We can measure the **Mean absolute error** by using this formulas [error=actual−predicted] . It will allow us to see how good our model is but for us to get actual predicted values on seprate data we will use ski learn's tool which will split the data in two parts, first which we will train our data on and second for testing our predictions .

```python
#importing mean_absolute_error from .metrics folder of sklearn
from sklearn.metrics import mean_absolute_error

predicted_prices = melbourne_model.predict(X)
mean_absolute_error(y,predicted_prices)
```

## What i did above was 'in sample' test which means we used the same sample to train the data and test the data , we do not want that , nooo bad thing , boooo . A very simple solution is the on ei talked above which is to seprate some data for later to test on . Btw this shi is called "DATA VALIDATION" .

```python
#importing train_test_split from .model selection folder of sklearn
from sklearn.model_selection import train_test_split

#we need to split both X and y in 2 parts each one for training and other for validation
train_X , val_X , train_y , val_y = train_test_split(X, y , random_state=0)

#now we can mfit our model with train data and test our model with validation dataframe
melbourne_model.fit(train_X,train_y)

melbourne_model.predict(val_X)
```

## Now you must be wondering why we created the val_y dataframe if we just needed the val_X to predict the model. That's where the DATA VALIDATION comes from , we are gonna use the mean absolute error on this new model to see it's accuracy

```python
val_predictions = melbourne_model.predict(val_X)
print(f'bruh look at this fuck ass model with this much error = {mean_absolute_error(val_y, val_predictions)} , totally unusable')
```

## The reason why our MAE came out so big coz the tree grew too much , probably overfittedd . which means it created hyper speciliased nodes for some very small amount of houses . we do not want that . so were gonna limit the amount of nodes it can generate to find the best leaf node size.

```python
'''We will create a small function here with all the normal arguments to create and data validate a model but we will also add another variable named max_leaf_nodes which will limit how many 
nodes can our model form , we will do trial method . keep changing the max leaf values each time to see which one gives the best MAE or lowest i mean .'''

def get_mae(max_leaf_nodes , train_X , val_X , train_y , val_y):
    new_model = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes , random_state=0)
    new_model.fit(train_X,train_y)
    pred_val = new_model.predict(val_X)
    mae = mean_absolute_error(val_y,pred_val)
    return mae

# Now we call the function

for i in [5, 50 , 100 , 200 , 250 , 500]:
    my_mae = get_mae(i , train_X , val_X , train_y , val_y)
    print(f'Max_leaf_nodes :{i} \t MAE = {my_mae}')
```

## Now that we know which amount of leaf node causes the least error we can just use that one in our model and don't even need to split data



```python
melbourne_model = DecisionTreeRegressor(max_leaf_nodes=250,random_state=0)
melbourne_model.fit(X,y)

my_mae = mean_absolute_error(y , melbourne_model.predict(X))

print(f'Now when you look at this MAE = {my_mae} , even though its so much bigger error than the first model we built with in sample test . Its still a superior model ')
```
